```python
from collections import Counter
import numpy as np

def plot_power_law(G, title=''):
    degrees = [d for _, d in G.degree()]
    degree_count = Counter(degrees)
    k = sorted(degree_count)
    pk = [degree_count[i] / sum(degree_count.values()) for i in k]

    plt.figure(figsize=(6, 4))
    plt.plot(k, pk, marker='o', linestyle='none', color='green')
    plt.xscale('log')
    plt.yscale('log')
    plt.xlabel('Degree k log')
    plt.ylabel('P(k) log')
    plt.title(title)
    plt.grid(True)
    plt.show()
```

```python
G = nx.barabasi_albert_graph(1000, 1)
plot_power_law(G, title='Barabási–Albert (BA)')

G = nx.erdos_renyi_graph(1000, 0.05)
plot_power_law(G, title='Erdős–Rényi (ER)')

G = nx.watts_strogatz_graph(1000, 20, 0.4)
plot_power_law(G, title='Watts–Strogatz (WS)')
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

ba_coeffs, er_coeffs, ws_coeffs = [], [], []

for _ in range(100):
    G_ba = nx.barabasi_albert_graph(1000, 2)
    ba_coeffs.append(nx.average_clustering(G_ba))

    G_er = nx.erdos_renyi_graph(1000, 0.004)
    er_coeffs.append(nx.average_clustering(G_er))

    G_ws = nx.watts_strogatz_graph(1000, 4, 0.1)
    ws_coeffs.append(nx.average_clustering(G_ws))

data = pd.DataFrame({
    'Clustering Coefficient': ba_coeffs + er_coeffs + ws_coeffs,
    'Model': ['Barabási–Albert'] * 100 + ['Erdős–Rényi'] * 100 + ['Watts–Strogatz'] * 100
})

sns.set(style='whitegrid', font_scale=1.2)
plt.figure(figsize=(6, 8))
palette = {'Barabási–Albert': '#6a0dad', 'Erdős–Rényi': '#1f77b4', 'Watts–Strogatz': '#2ca02c'}

ax = sns.boxplot(x='Model', y='Clustering Coefficient', data=data, palette=palette, width=0.5, fliersize=3)

medians = data.groupby('Model')['Clustering Coefficient'].median()
for i, model in enumerate(medians.index):
    median_val = medians[model]
    ax.text(i, median_val + 0.01, f'{median_val:.3f}', ha='center', va='bottom', color='black', fontsize=12)

plt.title('Distribution of Clustering Coefficients (100 Graphs per Model)', fontsize=16, weight='bold')
plt.xlabel('Graph Model', fontsize=14)
plt.ylabel('Clustering Coefficient', fontsize=14)
plt.ylim(-0.3, 0.6)
plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ba_path, er_path, ws_path = [], [], []

for _ in range(100):
    G_ba = nx.barabasi_albert_graph(1000, 2)
    if nx.is_connected(G_ba):
        ba_path.append(nx.average_shortest_path_length(G_ba))
    G_ws = nx.watts_strogatz_graph(1000, 4, 0.1)
    if nx.is_connected(G_ws):
        ws_path.append(nx.average_shortest_path_length(G_ws))

data = pd.DataFrame({
    'Average Shortest Path Length': ba_path + er_path + ws_path,
    'Model': ['Barabási–Albert'] * len(ba_path) +
             ['Erdős–Rényi'] * len(er_path) +
             ['Watts–Strogatz'] * len(ws_path)
})

sns.set(style='whitegrid', font_scale=1.2)
plt.figure(figsize=(6, 8))
palette = {'Barabási–Albert': '#6a0dad', 'Erdős–Rényi': '#1f77b4', 'Watts–Strogatz': '#2ca02c'}

ax = sns.boxplot(x='Model', y='Average Shortest Path Length', data=data, palette=palette, width=0.5, fliersize=3)

if not data.empty:
    medians = data.groupby('Model')['Average Shortest Path Length'].median()
    for i, model in enumerate(medians.index):
        median_val = medians[model]
        ax.text(i, median_val + 0.02, f'{median_val:.3f}', ha='center', va='bottom', color='black', fontsize=12)

plt.title('Distribution of Average Shortest Path Length (Connected Graphs)', fontsize=16, weight='bold')
plt.xlabel('Graph Model', fontsize=14)
plt.ylabel('Average Shortest Path Length', fontsize=14)
plt.tight_layout()
plt.show()
```
