```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import random
```

```python
#create cities and connections
citynet = nx.Graph()
citynet.add_nodes_from(["Almaty", "Shymkent", "Astana", "Kokshetau", "Semey", "Pavlodar"])
citynet.add_edges_from([
  ('Pavlodar', 'Astana'), 
  ('Pavlodar', 'Almaty'), 
  ("Almaty", "Astana"), 
  ("Almaty", "Shymkent"), 
  ("Kokshetau", 'Semey'), 
  ('Semey', 'Pavlodar')
  ])


#choose epidemic's parameters
b = 2
y = 1

#initialization of epidemic
reds = set()
blues = set(citynet.nodes)
blue_ns = list()
time = 1
cs0 = random.choice(list(citynet.nodes()))
print(cs0)
reds.add(cs0)
blues.remove(cs0)
```

```python
for node in citynet.nodes():
  if node in reds:
    citynet.nodes[node]['color'] = 'red'
  else:
    citynet.nodes[node]['color'] = 'blue'
node_colors = [citynet.nodes[node]['color'] for node in citynet.nodes]
nx.draw(citynet, with_labels = True, node_color = node_colors , node_size = 500)
```

```python
while True:
  blue_ns = []
  for red in reds:
    for neighbor in citynet.neighbors(red):
      if neighbor in blues:
        blue_ns.append(neighbor)
  ne = len(blue_ns)
  i = len(reds)
  pi = b * ne / (b * ne + i * y)
  pr = y * i / (b * ne + i * y)
  chance = random.random()
  print(pi, pr, chance)
  time += 1
  if chance <= pi:
    new_red = random.choice(blue_ns)
    citynet.nodes[new_red]['color'] = 'red'
    reds.add(new_red)
    print(blues, new_red)
    blues.remove(new_red)
    blue_ns.remove(new_red)
    
  else:
    new_green = random.choice(list(reds))
    citynet.nodes[new_green]['color'] = 'green'
    reds.remove(new_green)
  plt.figure()
  node_colors = [citynet.nodes[node]['color'] for node in citynet.nodes]
  nx.draw(citynet, with_labels = True, node_color = node_colors , node_size = 500)
  plt.show()
  if len(reds) == 0:
   break
print(time)
```
